```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

import warnings
warnings.filterwarnings("ignore")

print("Libraries imported successfully!")
```

```python
# Load the dataset
df = pd.read_excel(
    "../data/raw/default of credit card clients.xls",
    header=1
)

# Display the first five rows
df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

```python
target = "default payment next month"

# Count the number of each outcome
print(df[target].value_counts())

# Calculate percentages
print("\nTarget Distribution (%):")
print(df[target].value_counts(normalize=True) * 100)
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x=target
)

plt.title("Distribution of Credit Card Default")
plt.xlabel("Default Payment Next Month")
plt.ylabel("Number of Customers")
plt.xticks([0, 1], ["No Default", "Default"])

plt.show()
```

```python
# Separate features and target
X = df.drop(columns=["ID", target])
y = df[target]

print("Features Shape:", X.shape)
print("Target Shape:", y.shape)

print("\nTarget Distribution:")
print(y.value_counts())
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Set:", X_train.shape)
print("Testing Set:", X_test.shape)

print("\nTraining Target Distribution:")
print(y_train.value_counts(normalize=True))

print("\nTesting Target Distribution:")
print(y_test.value_counts(normalize=True))
```

```python

# Define categorical and numerical columns
categorical_cols = [
    "SEX",
    "EDUCATION",
    "MARRIAGE",
    "PAY_0",
    "PAY_2",
    "PAY_3",
    "PAY_4",
    "PAY_5",
    "PAY_6"
]

numerical_cols = [
    col for col in X.columns
    if col not in categorical_cols
]

# Apply one-hot encoding
X_train_encoded = pd.get_dummies(
    X_train,
    columns=categorical_cols,
    drop_first=True,
    dtype=int
)

X_test_encoded = pd.get_dummies(
    X_test,
    columns=categorical_cols,
    drop_first=True,
    dtype=int
)

# Make sure training and testing have the same columns
X_test_encoded = X_test_encoded.reindex(
    columns=X_train_encoded.columns,
    fill_value=0
)

# Standardize numerical variables
scaler = StandardScaler()

X_train_encoded[numerical_cols] = scaler.fit_transform(
    X_train_encoded[numerical_cols]
)

X_test_encoded[numerical_cols] = scaler.transform(
    X_test_encoded[numerical_cols]
)

print("Preprocessing completed!")
print("Processed Training Set:", X_train_encoded.shape)
print("Processed Testing Set:", X_test_encoded.shape)
```

```python
print("Final Training Data:")
print(X_train_encoded.head())

print("\nFinal Testing Data:")
print(X_test_encoded.head())

print("\nMissing Values in Training Data:")
print(X_train_encoded.isnull().sum().sum())

print("\nMissing Values in Testing Data:")
print(X_test_encoded.isnull().sum().sum())
```

## Bayesian Logistic Regression

For this project, we will first use Bayesian Logistic Regression to predict whether a credit card customer will default next month. The model will use the customer information and repayment history in the dataset to estimate the probability of default.

After fitting the model, we will check how well its predictions match the actual results in the test set. Since the dataset contains fewer default cases than non-default cases, we will also pay attention to how well the model identifies customers who actually default.

```python
# Bayesian Logistic Regression

import pymc as pm
from scipy.special import expit
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report,
    ConfusionMatrixDisplay,
    RocCurveDisplay
)

# Convert data to NumPy arrays
X_train_np = X_train_encoded.to_numpy(dtype=float)
X_test_np = X_test_encoded.to_numpy(dtype=float)

y_train_np = y_train.to_numpy(dtype=int)
y_test_np = y_test.to_numpy(dtype=int)

# Build the Bayesian Logistic Regression model
with pm.Model() as bayesian_logistic_model:

    # Prior distributions
    alpha = pm.Normal("alpha", mu=0, sigma=2.5)
    beta = pm.Normal(
        "beta",
        mu=0,
        sigma=1,
        shape=X_train_np.shape[1]
    )

    # Logistic regression
    logits = alpha + pm.math.dot(X_train_np, beta)

    # Bernoulli likelihood
    y_obs = pm.Bernoulli(
        "y_obs",
        logit_p=logits,
        observed=y_train_np
    )

    # Approximate posterior inference using ADVI
    approx = pm.fit(
        n=15000,
        method="advi",
        random_seed=42,
        progressbar=True
    )

    # Draw samples from the approximate posterior
    idata = approx.sample(1000, random_seed=42)

print("Bayesian Logistic Regression completed!")
```

```python
# Extract posterior samples
alpha_samples = np.asarray(
    idata.posterior["alpha"]
).reshape(-1)

beta_samples = np.asarray(
    idata.posterior["beta"]
).reshape(-1, X_train_np.shape[1])

# Calculate predicted probabilities for the test set
logits_test = (
    alpha_samples[:, None]
    + beta_samples @ X_test_np.T
)

probability_samples = expit(logits_test)

# Average posterior probabilities
y_pred_proba = probability_samples.mean(axis=0)

# Convert probabilities into class predictions
y_pred = (y_pred_proba >= 0.5).astype(int)

print("Prediction completed!")
print("Number of predictions:", len(y_pred))
```

```python
accuracy = accuracy_score(y_test_np, y_pred)
precision = precision_score(y_test_np, y_pred, zero_division=0)
recall = recall_score(y_test_np, y_pred, zero_division=0)
f1 = f1_score(y_test_np, y_pred, zero_division=0)
auc = roc_auc_score(y_test_np, y_pred_proba)

print("Bayesian Logistic Regression Results")
print("------------------------------------")
print(f"Accuracy:  {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall:    {recall:.4f}")
print(f"F1 Score:  {f1:.4f}")
print(f"ROC-AUC:   {auc:.4f}")

print("\nClassification Report:")
print(classification_report(
    y_test_np,
    y_pred,
    target_names=["No Default", "Default"],
    zero_division=0
))
```

```python
# Confusion Matrix
fig, ax = plt.subplots(figsize=(6, 5))

ConfusionMatrixDisplay.from_predictions(
    y_test_np,
    y_pred,
    display_labels=["No Default", "Default"],
    cmap="Blues",
    ax=ax
)

plt.title("Bayesian Logistic Regression - Confusion Matrix")
plt.show()


# ROC Curve
fig, ax = plt.subplots(figsize=(6, 5))

RocCurveDisplay.from_predictions(
    y_test_np,
    y_pred_proba,
    ax=ax
)

plt.title("Bayesian Logistic Regression - ROC Curve")
plt.show()
```

## Coefficient interpretation

Each coefficient shows how a feature changes the log-odds of default, holding the others fixed. A positive value raises the predicted risk and a negative value lowers it. `odds_multiplier` (e^coefficient) says how much the odds of default are multiplied when that feature goes up by one unit. For the standardized continuous features, one unit is one standard deviation. For the one-hot features, it means being in that category instead of the baseline category that was dropped (for the PAY_ columns, the baseline is -2, meaning no balance to pay that month). If the 95% credible interval doesn't cross 0, the model is fairly confident about the direction of the effect.

```python
# Posterior summary for every coefficient: mean and 95% credible interval
coef_summary = pd.DataFrame({
    "feature": X_train_encoded.columns,
    "mean": beta_samples.mean(axis=0),
    "lower_95": np.percentile(beta_samples, 2.5, axis=0),
    "upper_95": np.percentile(beta_samples, 97.5, axis=0),
})
coef_summary["odds_multiplier"] = np.exp(coef_summary["mean"])
coef_summary["interval_excludes_0"] = (coef_summary["lower_95"] > 0) | (coef_summary["upper_95"] < 0)
coef_summary.to_csv("../figures/bayes_coefficients.csv", index=False)

# Plot the 15 coefficients with the largest effect (by absolute posterior mean)
top = coef_summary.loc[coef_summary["mean"].abs().sort_values(ascending=False).index].head(15).iloc[::-1]

fig, ax = plt.subplots(figsize=(7, 6))
ax.errorbar(top["mean"], range(len(top)),
            xerr=[top["mean"] - top["lower_95"], top["upper_95"] - top["mean"]],
            fmt="o", capsize=3)
ax.axvline(0, color="gray", linestyle="--")
ax.set_yticks(range(len(top)))
ax.set_yticklabels(top["feature"])
ax.set_xlabel("Coefficient (log-odds of default)")
ax.set_title("Bayesian Logistic Regression: top 15 coefficients\n(posterior mean with 95% credible interval)")
fig.tight_layout()
fig.savefig("../figures/bayes_coefficients.png", dpi=200)
plt.show()

coef_summary.sort_values("mean", key=abs, ascending=False).head(15)
```
